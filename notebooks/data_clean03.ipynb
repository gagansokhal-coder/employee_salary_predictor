```python
import ast
import pandas as pd

def safe_parse(x):
    if pd.isna(x):
        return {}

    if isinstance(x, dict):
        return x

    if isinstance(x, str) and x.strip().startswith("{"):
        try:
            return ast.literal_eval(x)
        except Exception:
            return {}

    return {}

df["company"] = df["company"].apply(safe_parse)
```

```python
print(type(df.loc[1, "company"]))
print(df.loc[1, "company"])
```

```python
df["company_name"] = df["company"].apply(
    lambda x: x.get("display_name")
)
```

```python
df['category']=df['category'].apply(safe_parse)
```

```python
df['category_name']=df['category'].apply(
    lambda x:x.get('label')
)
df['category_tag']=df['category'].apply(
    lambda x: x.get('tag')
)
```

```python
df['location']=df['location'].apply(safe_parse)
```

```python
df['location_name']=df['location'].apply(
    lambda x:x.get('display_name')
)
```

```python
df['location_area']=df['location'].apply(
    lambda x:x.get('area')
)
```

```python
df[['location_area','location_name']].head(2)
```

```python
df['country']=df['location_area'].apply(
    lambda x: x[0] if isinstance(x,list) and len(x)>0 else None
)
df['State']=df['location_area'].apply(
    lambda x:x[1] if isinstance(x,list) and len(x)>1 else None
)
df["county"] = df["location_area"].apply(
    lambda x: x[2] if isinstance(x, list) and len(x) > 2 else None
)

df["city"] = df["location_area"].apply(
    lambda x: x[-1] if isinstance(x, list) and len(x) > 0 else None
)
```

```python
df[
    [
        "company_name",
        "category_name",
        "category_tag",
        "country",
        "State",
        "county",
        "city"
    ]
].head()
```

```python
df.drop(
    columns=[
        "company",
        "category",
        "location"
    ],
    inplace=True
)
```
