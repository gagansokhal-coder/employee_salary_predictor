```python
import ast
import pandas as pd

def safe_parse(x):
    if pd.isna(x):
        return {}

    if isinstance(x, dict):
        return x

    if isinstance(x, str) and x.strip().startswith("{"):
        try:
            return ast.literal_eval(x)
        except Exception:
            return {}

    return {}

df["company"] = df["company"].apply(safe_parse)
```

```python
df['company'].isnull().mean()*100
```

```python
df['company_name']=df['company'].apply(
    lambda x: x.get('display_name')
)
```
